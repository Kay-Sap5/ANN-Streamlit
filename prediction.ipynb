```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import numpy as np
import pandas as pd
import pickle
```

```python
model = load_model("model.h5")

with open("label_encoder_gender.pkl",'rb') as file:
    label_encoder_gender = pickle.load(file)

with open("onehotencoder_geo.pkl" , 'rb') as file:
    ohe_geo = pickle.load(file)

with open("scaler.pkl",'rb') as file:
    scaler = pickle.load(file)
```

```python
input_data = {'CreditScore': 500,
 'Geography': 'France',
 'Gender': 'Male',
 'Age': 30,
 'Tenure': 2,
 'Balance': 50000,
 'NumOfProducts': 1,
 'HasCrCard': 1,
 'IsActiveMember':1,
 'EstimatedSalary':60000,}
```

```python
new_df  = pd.DataFrame([input_data])
```

```python
new_df
```

```python
new_df['Gender']  = label_encoder_gender.transform([new_df['Gender']])
```

```python
new_df
```

```python
geo_df = pd.DataFrame(data=ohe_geo.transform([new_df['Geography']]).toarray() , columns=ohe_geo.get_feature_names_out())
```

```python
geo_df
```

```python
new_df
```

```python
final_df = pd.concat([new_df.drop(columns=['Geography']) , geo_df ],axis=1)
```

```python
new_df.drop(columns=['Geography'])
```

```python
final_df = scaler.fit_transform(final_df)
```

```python
model.predict(final_df)
```
